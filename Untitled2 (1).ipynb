```python
import matplotlib.pyplot as plt

df.groupby('Category')['Amount'].sum().sort_values().plot(kind='barh')
plt.title('Sales by Category')
plt.xlabel('Sales Amount')
plt.ylabel('Category')
plt.show()
```

```python
df.groupby(df['Date'].dt.to_period('M'))['Amount'].sum().plot(kind='bar')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
plt.show()
```

```python
df.groupby('ship-state')['Amount'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title('Top 10 States by Sales')
plt.xlabel('State')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby('Category')['Amount'].sum().sort_values().plot(kind='barh')
plt.title('E-Commerce Sales by Category')
plt.xlabel('Sales Amount (INR)')
plt.show()
```

```python
df.groupby('Fulfilment')['Amount'].sum().plot(kind='bar')
plt.title('Sales by Fulfilment')
plt.xlabel('Fulfilment')
plt.ylabel('Sales Amount (INR)')
plt.show()
```

```python
df.groupby('ship-state')['Amount'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title('Top 10 States by Sales')
plt.xlabel('State')
plt.ylabel('Sales Amount (INR)')
plt.xticks(rotation=45)
plt.show()
```

```python
df['Status'].value_counts().head(8).plot(kind='bar')
plt.title('Order Status Distribution')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
print("Total Orders:", len(df))
print("Total Sales:", df['Amount'].sum())
print("Average Order Value:", df['Amount'].mean())
print("Cancellation Rate:", df['Status'].eq('Cancelled').mean() * 100)
```
